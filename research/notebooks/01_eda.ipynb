```python
from ucimlrepo import fetch_ucirepo

datasets = [
    ("Bank Marketing", 222),
    ("Adult", 2),
]

for name, dataset_id in datasets:
    dataset = fetch_ucirepo(id=dataset_id)
    X = dataset.data.features
    y = dataset.data.targets.iloc[:, 0]

    print(f"\n--- {name} ---")
    print("件数:", len(X))
    print("特徴量数:", X.shape[1])
    print("欠損値（空欄）:", int(X.isna().sum().sum()))
    print("欠損値（?表記）:", int((X.astype("string") == "?").sum().sum()))
    print("クラスごとの件数:")
    print(y.value_counts(dropna=False))
    print("最少クラスの件数:", int(y.value_counts().min()))
```
